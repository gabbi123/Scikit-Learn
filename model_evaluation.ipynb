```python
# Evaluation 1
# Train the model on the entire dataset.
# Test the model on the same dataset, and evaluate how well 
# we did by comparing the predicted response values with the true response values.
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
# features and response
X = iris.data
y = iris.target
print (y)
```

```python
# Linear regression
from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(solver = 'liblinear')

# fit the model with the data
logreg.fit(X, y)

# predict for the same datasset
y_pred = logreg.predict(X)

print(y_pred.shape)
```

```python
# compute classification accuracy
from sklearn import metrics

print(metrics.accuracy_score(y, y_pred))
```

```python
# Known as training accuracy when you train and test the model on the same data
```

```python
# KNN (N = 5)
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors = 5)

knn.fit(X, y)

y_pred_knn5 = knn.predict(X)

print(metrics.accuracy_score(y, y_pred_knn5))
```

```python
# KNN (N = 1)
from sklearn.neighbors import KNeighborsClassifier

knn_1 = KNeighborsClassifier(n_neighbors = 1)

knn_1.fit(X, y)

y_pred_knn1 = knn_1.predict(X)

print(metrics.accuracy_score(y, y_pred_knn1))
```

```python
# Problems with training and testing on the same data
# Overfiting
```

```python
# Evaluation 2
# Train/test split
# Split the dataset into two pieces: a training set and a testing set.
```

```python
print(X.shape)
print(y.shape)
```

```python
# train test sample set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.4, random_state = 4)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
print(y_train.shape)
print(y_test.shape)
```

```python
# logical regression with training set. Get the accuracy with the testing dataset
log_reg = LogisticRegression(solver='liblinear')

log_reg.fit(X_train, y_train)

pred_y_test = log_reg.predict(X_test)

print(metrics.accuracy_score(y_test, pred_y_test))
```

```python
# KNN (N = 5) with train and test dataset
knn_5 = KNeighborsClassifier(n_neighbors = 5)

knn_5.fit(X_train, y_train)

pred_knn_y_test = knn_5.predict(X_test)

print(metrics.accuracy_score(y_test, pred_knn_y_test))
```

```python
# KNN (N = 1) with training and test dataset
knn_1 = KNeighborsClassifier(n_neighbors = 1)

knn_1.fit(X_train, y_train)

pred_knn_1_y_test = knn_1.predict(X_test)

print(metrics.accuracy_score(y_test, pred_knn_1_y_test))
```

```python
#Locate better N value - KNN from 1 to 25
import numpy as np
import matplotlib.pyplot as plt

n_range = np.arange(1,26)
accuracy_scores = []

for n_value in n_range:
    knn = KNeighborsClassifier(n_neighbors = n_value)
    knn.fit(X_train, y_train)

    y_score = knn.predict(X_test)
    accuracy_scores.append(metrics.accuracy_score(y_test, y_score))

%matplotlib inline

plt.plot(n_range, accuracy_scores)
plt.xlabel('K value of KNN')
plt.ylabel('accuracy score')
```

```python
# Make prediction for out of sample data
knn = KNeighborsClassifier(n_neighbors = 11)

knn.fit(X, y)

print(knn.predict([[3, 4, 5, 2]]))
```
